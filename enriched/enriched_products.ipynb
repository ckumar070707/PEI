```python
def enriched_products(transform_path,enriched_path):
    try:

        df_transform = spark.read.format("delta").load(transform_path)
        col_list_select = ['Product_ID','Category','Sub-Category']
        df_transform = df_transform.select(col_list_select).dropDuplicates()
        df_transform = df_transform.fillna('Unknown')
        df_transform.write.format("delta").mode("overwrite").option("overwriteSchema", "true").save(enriched_path)
    except Exception as e:
        print(f"someting went wrong {e}")
```

```python
transform_path = "/Volumes/workspace/default/transformed/Products"
enriched_path = "/Volumes/workspace/default/enriched/Products"
enriched_products(transform_path,enriched_path)
```
