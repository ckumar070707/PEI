```python
def Orders_transform(raw_path,transform_path):
    try:
        df_raw = spark.read.option("multiLine", True).json(raw_path)
        original_cols = df_raw.columns
        cleaned_cols = [c.replace(' ', '_')for c in original_cols ]
        #todo data check
        df_raw = df_raw.toDF(*cleaned_cols)
        df_raw.write.format("delta").mode("overwrite").option("overwriteSchema", "true").save(transform_path)
    except Exception as e:
        print(f"someting went wrong {e}")
```

```python
raw_path = '/Volumes/workspace/default/raw/Orders.json'
transform_path = "/Volumes/workspace/default/transformed/Orders"
Orders_transform(raw_path,transform_path)
```
