```python
def Products_transform(raw_path,transform_location,):
    try:
        raw_path = '/Volumes/workspace/default/raw/Products.csv'
        transform_location = "/Volumes/workspace/default/transformed/Products"
        df_raw = spark.read.format("csv").option("header", "true").option("inferSchema","True").option("escape",'"').load(raw_path)
        original_cols = df_raw.columns
        cleaned_cols = [c.replace(' ', '_')for c in original_cols ]
        
        #todo data check
        df_raw = df_raw.toDF(*cleaned_cols)
        df_raw.write.format("delta").mode("overwrite").option("overwriteSchema", "true").save(transform_location)
    

    except Exception as e:
        print(f"Error occured during transformation: {e}")
```
