```python
from pyspark.sql import functions as F
```

```python
def agg_profit(enriched_orders_path,agg_profit_path):
    try:

        df_orders = spark.read.format("delta").load(enriched_orders_path)
        df_orders = df_orders.withColumn("Year",F.year("Order_Date").cast("int"))
        agg_col_list = ['Year','Product_Category','Product_Sub_Category','Customer_Name']
        df_agg = df_orders.groupBy(agg_col_list)\
                          .agg( 
                               F.round(F.sum("Profit"),2).alias("Total_Profit")
                               )

        

        df_agg.write.format("delta").mode("overwrite").option("overwriteSchema", "true").save(agg_profit_path)
    except Exception as e:
        print(f"someting went wrong {e}")
```

```python
enriched_orders_path = "/Volumes/workspace/default/enriched/Orders"
agg_profit_path = "/Volumes/workspace/default/enriched/Profit"
agg_profit(enriched_orders_path,agg_profit_path)
```
