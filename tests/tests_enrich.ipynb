```python
import pytest
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, IntegerType, LongType
```

```python
def test_gold_layer_schema(spark):
    df = spark.table("gold_layer_table")
    expected_columns = {"id", "name", "value", "timestamp"}
    assert set(df.columns) == expected_columns

def test_gold_layer_row_count(spark):
    df = spark.table("gold_layer_table")
    assert df.count() > 0

def test_gold_layer_value_range(spark):
    df = spark.table("gold_layer_table")
    min_value = df.agg({"value": "min"}).collect()[0][0]
    max_value = df.agg({"value": "max"}).collect()[0][0]
    assert min_value >= 0
    assert max_value <= 100

def test_gold_layer_no_nulls(spark):
    df = spark.table("gold_layer_table")
    null_counts = df.select([df[c].isNull().cast("int").alias(c) for c in df.columns]).groupBy().sum().collect()[0]
    assert all(count == 0 for count in null_counts)
```
